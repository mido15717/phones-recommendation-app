```python
import pandas as pd

df = pd.read_csv("../src/data/raw_data/phone_with_specs.csv")

df.head()
```

```python
def get_unique_keys(df):
    unique_keys = set()

    for text in df["key_features_text"].dropna():
        pairs = text.split(",")

        for pair in pairs:
            if ":" in pair:
                key, value = pair.split(":", 1)
                unique_keys.add(key.strip())

    return sorted(unique_keys)
```

```python
df2 = get_unique_keys(df)
print(f"Unique keys: {df2}")
```

```python
def parse_features(df):
    """
    This function takes a dataframe and parses the 'key_features_text' column into a list of features.
    It returns a new dataframe with an additional column 'parsed_features' containing the list of features.
    """
    df['parsed_features'] = df['key_features_text'].apply(lambda x: x.split(',') if pd.notnull(x) else [])
    return df
```
